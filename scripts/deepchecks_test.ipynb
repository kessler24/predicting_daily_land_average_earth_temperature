```python
import pandas as pd
from deepchecks.tabular import Dataset
# ----- Create a dummy dataset -----
data = {
    'age': [25, 30, 45, 50, 60],
    'tumor_size': [1.2, 3.4, 2.5, 5.6, 7.8],
    'gender': ['M', 'F', 'F', 'M', 'F'],
    'class': [0, 1, 0, 1, 1]   # label column
}
dummy_df = pd.DataFrame(data)
# ----- Specify categorical features -----
categorical_columns = ['gender']
# ----- Create the Deepchecks Dataset object -----
dummy_ds = Dataset(
    dummy_df,
    label="class",
    cat_features=categorical_columns
)
dummy_ds
```
